```python
from pathlib import Path
import pandas as pd

project = Path.home() / "Documents" / "BSE_Climate_Policy"

countries = [
    "AUS", "CAN", "DEU", "ESP", "FRA", "GBR",
    "IRL", "ITA", "MEX", "NZL", "USA"
]

# 11개국 × 2000–2019년
carbon_panel = pd.MultiIndex.from_product(
    [countries, range(2000, 2020)],
    names=["country_code", "year"]
).to_frame(index=False)

# EU ETS 적용국: 2005년부터
eu_countries = ["DEU", "ESP", "FRA", "GBR", "IRL", "ITA"]

carbon_panel["CarbonPrice_draft"] = 0.0

carbon_panel.loc[
    carbon_panel["country_code"].isin(eu_countries)
    & (carbon_panel["year"] >= 2005),
    "CarbonPrice_draft"
] = 1.0

# 그 밖의 국가 단위 제도
carbon_panel.loc[
    (carbon_panel["country_code"] == "NZL")
    & (carbon_panel["year"] >= 2008),
    "CarbonPrice_draft"
] = 1.0

carbon_panel.loc[
    (carbon_panel["country_code"] == "MEX")
    & (carbon_panel["year"] >= 2014),
    "CarbonPrice_draft"
] = 1.0

carbon_panel.loc[
    (carbon_panel["country_code"] == "CAN")
    & (carbon_panel["year"] >= 2019),
    "CarbonPrice_draft"
] = 1.0

# 호주 CPM: 2012–2013년 시행, 2014년은 연중 폐지되어 별도 검토
carbon_panel.loc[
    (carbon_panel["country_code"] == "AUS")
    & (carbon_panel["year"].isin([2012, 2013])),
    "CarbonPrice_draft"
] = 1.0

carbon_panel.loc[
    (carbon_panel["country_code"] == "AUS")
    & (carbon_panel["year"] == 2014),
    "CarbonPrice_draft"
] = float("nan")

carbon_panel
```

```python
carbon_panel[
    carbon_panel["year"].isin([2004, 2005, 2008, 2012, 2013, 2014, 2015, 2019])
].pivot(
    index="country_code",
    columns="year",
    values="CarbonPrice_draft"
)
```

```python
carbon_panel.to_csv(
    project / "processed_data" / "carbon_price_panel_draft_2000_2019.csv",
    index=False
)
```

```python
cpu_panel = pd.read_csv(
    project / "processed_data" / "cpu_long_2000_2019.csv"
)

carbon_panel = pd.read_csv(
    project / "processed_data" / "carbon_price_panel_draft_2000_2019.csv"
)

panel = cpu_panel.merge(
    carbon_panel,
    on=["country_code", "year"],
    how="inner",
    validate="one_to_one"
)

print("행·열 수:", panel.shape)
print("국가 수:", panel["country_code"].nunique())
print("중복 국가·연도:", panel.duplicated(["country_code", "year"]).sum())
print("CPU 결측:", panel["CPU"].isna().sum())
print("탄소가격 변수 결측:", panel["CarbonPrice_draft"].isna().sum())

panel.head()
```
